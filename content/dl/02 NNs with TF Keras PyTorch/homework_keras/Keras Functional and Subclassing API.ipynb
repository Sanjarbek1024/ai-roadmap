```python
# Building Regression Model using Keras Sequential API
from tensorflow.keras.datasets import california_housing
```

```python
housing = california_housing.load_data()
```

```python
(X_train, y_train), (X_test, y_test) = housing
```

```python
X_train.shape
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Normalization, InputLayer
```

```python
norm_layer = Normalization(name='norm_layer')

model = Sequential([
    InputLayer(shape=[8], name='input_layer'),
    norm_layer,
    Dense(100, activation='relu', name='hidden1'),
    Dense(100, activation='relu', name='hidden2'),
    Dense(100, activation='relu', name='hidden3'),
    Dense(1, name='output') # output layer
])
```

```python
model.summary()
```

```python
norm_layer.get_weights()
```

```python
norm_layer.adapt(X_train)
```

```python
norm_layer.get_weights()
```

```python
from tensorflow.keras.optimizers import SGD, Adam
```

```python
optimizer = Adam(learning_rate=0.001)
model.compile(optimizer=optimizer, loss='mse', metrics=['RootMeanSquaredError'])
```

```python
history = model.fit(X_train, y_train, epochs=50, validation_split=0.2)
```

```python
y_pred = model.predict(X_train)
```

```python
model.evaluate(X_train, y_train)
```

```python
from tensorflow.keras.metrics import R2Score
```

```python
scorer = R2Score()
```

```python
scorer.update_state(y_train, y_pred)
```

# Functional API

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense, Normalization, Input, Concatenate
```

```python
normalization_layer = Normalization()
hidden_layer1 = Dense(30, activation='relu')
hidden_layer2 = Dense(30, activation='relu')
concat_layer = Concatenate()
output_layer = Dense(1)

input_ = Input(shape=[8])

normalized = normalization_layer(input_)
hidden1 = hidden_layer1(normalized)
hidden2 = hidden_layer2(hidden1)

concatenated = concat_layer([input_, hidden2])

output = output_layer(concatenated)

model = Model(inputs=[input_], outputs=[output])
```

```python
model.summary()
```

```python
model.save('deep_and_wide_model.keras')
```

```python
input_layer_deep = Input([4])
input_layer_wide = Input([4])

normalization_layer_deep = Normalization(name='norm_deep')
normalization_layer_wide = Normalization(name='norm_wide')

hidden_layer1 = Dense(30, activation='relu')
hidden_layer2 = Dense(30, activation='relu')
concat_layer = Concatenate()
output_layer = Dense(1)

# ---------------------

normalized_deep = normalization_layer_deep(input_layer_deep)
normalized_wide = normalization_layer_wide(input_layer_wide)

hidden1 = hidden_layer1(normalized_deep)
hidden2 = hidden_layer2(hidden1)

concatenated = concat_layer([hidden2, normalized_wide])

output = output_layer(concatenated)

model = Model(inputs=[normalized_wide, input_layer_deep], outputs=[output, hidden2])
```

```python
normalization_layer_deep.adapt()
normalization_layer_wide.adapt()
```

```python
model.save('multiple_output_model.keras')
```

# Subclass API

```python
from tensorflow.keras.models import Model
```

```python
class MyModel(Model):
  def __init__(self, units=30, out=1, activation='relu', **kwargs):
    super().__init__(**kwargs)
    # self.input_layer = InputLayer(shape=shape)
    self.norm_layer = Normalization()
    self.hidden1 = Dense(units, activation=activation)
    self.hidden2 = Dense(units, activation=activation)
    self.out = Dense(1)

  def call(self, X):
    # X = self.input_layer(X)
    X = self.norm_layer(X)
    # print('here')
    X = self.hidden1(X)
    X = self.hidden2(X)
    X = self.out(X)
    return X

my_model = MyModel(units=50)
my_model.norm_layer.adapt(X_train)
```

```python
optimizer = Adam(learning_rate=0.001)
my_model.compile(optimizer=optimizer, loss='mse', metrics=['RootMeanSquaredError'])
history = my_model.fit(X_train, y_train, epochs=50, validation_split=0.2)
```

```python
my_model.save('custom.keras')
```
