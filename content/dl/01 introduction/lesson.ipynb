```python
class MLP_Class:
    def __init__(self, input_size: int = 3, output_size: int = 3):
        self.input_size = input_size
        self.output_size = output_size

    def predict(self, X):
        W = np.random.randn(self.input_size, self.output_size)
        b = np.random.randn(self.output_size)
        z = X @ W + b
        y = np.heaviside(z, 0)

        return y
    
    @property
    def parameter_size(self):
        return (self.input_size + 1) * self.output_size
    
mlp = MLP_Class(784, 10)
mlp.parameter_size
```

```python
class FNN: #Feedforward Neural Network
    def __init__(self, input_size: int, hidden_sizes: list[int], output_size: int):
        self.input_size = input_size
        self.hidden_sizes = hidden_sizes
        self.output_size = output_size

    def fit(self, X, y):
        layer_sizes = [self.input_size] + self.hidden_sizes + [self.output_size]
        Ws = []
        bs = []

        for size1, size2 in zip(layer_sizes, layer_sizes[1:]):
            W = np.random.randn(size1, size2)
            b = np.random.randn(size2)
            Ws.append(W)
            bs.append(b)

        # And we have to fit them

    def predict(self, X):
        layer_sizes = [self.input_size] + self.hidden_sizes + [self.output_size]

        h_in = X
        for size1, size2 in zip(layer_sizes, layer_sizes[1:]):
            W = np.random.randn(size1, size2)
            b = np.random.randn(size2)
            z = h_in @ W + b
            h_out = np.heaviside(z, 0)
            h_in = h_out

        return h_out


    @property
    def parameter_size(self):
        return (self.input_size + 1) * self.hidden_sizes[0] + sum((h + 1) * n for h, n in zip(self.hidden_sizes[:-1], self.hidden_sizes[1:])) + (self.hidden_sizes[-1] + 1) * self.output_size
    

nn = FNN(input_size=8, hidden_sizes=[6, 4, 2, 7], output_size=10)
nn.fit(X, y)
nn.predict(X)
```
