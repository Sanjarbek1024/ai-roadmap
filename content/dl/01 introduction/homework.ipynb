```python
import numpy as np
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing()

X = housing['data']
y = housing['target'].reshape(-1, 1)
```

```python
from sklearn.preprocessing import StandardScaler

scaler_x = StandardScaler()
scaler_y = StandardScaler()
X = scaler_x.fit_transform(X)
y = scaler_y.fit_transform(y)
```

```python
class FNN:  # Feedforward Neural Network
    def __init__(self, input_size: int, hidden_sizes: list[int], output_size: int, lr=0.01, epochs=500):
        self.input_size = input_size
        self.hidden_sizes = hidden_sizes
        self.output_size = output_size
        self.lr = lr
        self.epochs = epochs

    def _relu(self, x):
        return np.maximum(0, x)

    def _relu_deriv(self, x):
        return (x > 0).astype(float)

    def fit(self, X, y):
        layer_sizes = [self.input_size] + self.hidden_sizes + [self.output_size]
        self.Ws = []
        self.bs = []

        # initialize weights
        for size1, size2 in zip(layer_sizes, layer_sizes[1:]):
            W = np.random.randn(size1, size2) * np.sqrt(2 / size1)
            b = np.zeros((1, size2))
            self.Ws.append(W)
            self.bs.append(b)

        # gradient descent training
        for epoch in range(self.epochs):
            # === Forward ===
            a_values = [X]
            z_values = []
            a = X
            for W, b in zip(self.Ws[:-1], self.bs[:-1]):  # hidden layers
                z = a @ W + b
                a = self._relu(z)
                z_values.append(z)
                a_values.append(a)

            # output layer (linear)
            z = a @ self.Ws[-1] + self.bs[-1]
            a_values.append(z)
            z_values.append(z)
            y_pred = z

            # === Loss (MSE) ===
            loss = np.mean((y_pred - y) ** 2)

            # === Backward ===
            dA = 2 * (y_pred - y) / y.shape[0]
            for i in reversed(range(len(self.Ws))):
                Z = z_values[i]
                A_prev = a_values[i]
                if i < len(self.Ws) - 1:
                    dZ = dA * self._relu_deriv(Z)
                else:
                    dZ = dA  # last layer linear

                dW = A_prev.T @ dZ
                db = np.sum(dZ, axis=0, keepdims=True)
                if i > 0:
                    dA = dZ @ self.Ws[i].T

                # update weights
                self.Ws[i] -= self.lr * dW
                self.bs[i] -= self.lr * db

            if epoch % 100 == 0:
                print(f"Epoch {epoch} | Loss: {loss:.4f}")

    def predict(self, X):
        a = X
        for W, b in zip(self.Ws[:-1], self.bs[:-1]):
            z = a @ W + b
            a = self._relu(z)
        # output layer (linear)
        return a @ self.Ws[-1] + self.bs[-1]

    @property
    def parameter_size(self):
        return sum(W.size + b.size for W, b in zip(self.Ws, self.bs))

```

```python
nn = FNN(input_size=8, hidden_sizes=[6, 4, 2, 7], output_size=1, lr=0.01, epochs=1000)
nn.fit(X, y)
y_pred = nn.predict(X)
print("Final MSE:", np.mean((y_pred - y) ** 2))
print("Parameter size:", nn.parameter_size)
```
