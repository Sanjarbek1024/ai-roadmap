```python
import numpy as np

class PCAclass:
    def __init__(self, n_components):
        self.n_components = n_components
        self.components_ = None
        self.mean_ = None
        self.explained_variance_ = None
        self.explained_variance_ratio_ = None

    def fit(self, X):
        # 1. Center the data
        self.mean_ = np.mean(X, axis=0)
        X_centered = X - self.mean_

        # 2. Covariance matrix
        cov_matrix = np.cov(X_centered, rowvar=False)

        # 3. Eigen decomposition
        eig_vals, eig_vecs = np.linalg.eigh(cov_matrix)

        # 4. Sort eigenvalues (descending)
        sorted_idx = np.argsort(eig_vals)[::-1]
        eig_vals = eig_vals[sorted_idx]
        eig_vecs = eig_vecs[:, sorted_idx]

        # 5. Select first n_components
        self.components_ = eig_vecs[:, :self.n_components]

        # 6. Save explained variance info
        self.explained_variance_ = eig_vals[:self.n_components]
        self.explained_variance_ratio_ = eig_vals[:self.n_components] / np.sum(eig_vals)

    def transform(self, X):
        X_centered = X - self.mean_
        return np.dot(X_centered, self.components_)

    def fit_transform(self, X):
        self.fit(X)
        return self.transform(X)
```

```python
# Fake dataset
X = np.random.rand(100, 5)

pca = PCAclass(n_components=2)
X_reduced = pca.fit_transform(X)

print("Original shape:", X.shape)
print("Reduced shape:", X_reduced.shape)
print("Explained variance ratio:", pca.explained_variance_ratio_)
```

```python
import numpy as np

class PCAclassSVD:
    def __init__(self, n_components):
        self.n_components = n_components
        self.components_ = None
        self.mean_ = None
        self.explained_variance_ = None
        self.explained_variance_ratio_ = None

    def fit(self, X):
        # 1. Center the data
        self.mean_ = np.mean(X, axis=0)
        X_centered = X - self.mean_

        # 2. SVD
        U, S, Vt = np.linalg.svd(X_centered, full_matrices=False)

        # 3. Components (asosiy yo‘nalishlar)
        self.components_ = Vt[:self.n_components].T

        # 4. Explained variance (S^2 / (n_samples - 1))
        n_samples = X.shape[0]
        eig_vals = (S**2) / (n_samples - 1)

        self.explained_variance_ = eig_vals[:self.n_components]
        self.explained_variance_ratio_ = eig_vals[:self.n_components] / np.sum(eig_vals)

    def transform(self, X):
        X_centered = X - self.mean_
        return np.dot(X_centered, self.components_)

    def fit_transform(self, X):
        self.fit(X)
        return self.transform(X)
```

```python
# Fake dataset
X = np.random.rand(100, 5)

pca = PCAclassSVD(n_components=2)
X_reduced = pca.fit_transform(X)

print("Original shape:", X.shape)
print("Reduced shape:", X_reduced.shape)
print("Explained variance ratio:", pca.explained_variance_ratio_)
```
