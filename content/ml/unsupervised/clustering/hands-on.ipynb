```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
```

```python
X = pd.read_csv('data/2d_points.csv').values
X.shape
```

```python
centroids = np.random.rand(3, 2)
```

```python
centroids
```

```python
plt.scatter(X[:, 0], X[:, 1])
plt.scatter(centroids[:, 0], centroids[:, 1], color=['red', 'green', 'blue'], marker='X', s=20)
plt.show()
```

```python
c_0 = np.sqrt(np.sum((X - centroids[0])**2, axis=1))
c_1 = np.sqrt(np.sum((X - centroids[1])**2, axis=1))
c_2 = np.sqrt(np.sum((X - centroids[2])**2, axis=1))
```

```python
distance = np.stack((c_0, c_1, c_2), axis=1)
distance.shape
```

```python
clusters = distance.argmin(axis=1)
```

```python
plt.scatter(X[:, 0], X[:, 1], c=clusters, cmap='viridis')
plt.scatter(centroids[:, 0], centroids[:, 1], color='green', marker='X', s=20)
plt.show()
```
